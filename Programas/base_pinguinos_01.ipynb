```python
import matplotlib.pyplot as plt

# Cargar el conjunto de entrenamiento
train_data = pd.read_csv('/content/penguins_training.csv')

# Crear un gráfico de dispersión para las características 'Culmen Length (mm)' y 'Culmen Depth (mm)'
plt.figure(figsize=(10, 6))

# Colorear por especie
species = {'Adelie Penguin (Pygoscelis adeliae)'      : 'blue',
            'Chinstrap penguin (Pygoscelis antarctica)': 'green',
            'Gentoo penguin (Pygoscelis papua)'        : 'red'}

for specie, color in species.items():
    subset = train_data[train_data['Species'] == specie]
    plt.scatter(subset['Culmen Length (mm)'], subset['Culmen Depth (mm)'],
                color=color, label=specie, alpha=0.6)

# Agregar etiquetas y título
plt.xlabel('Culmen Length (mm)')
plt.ylabel('Culmen Depth (mm)')
plt.title('Comportamiento de Culmen Length vs Culmen Depth por Especie')
plt.legend()
plt.grid(True)

# Mostrar gráfico
plt.show()
```

```python
%matplotlib inline
from mpl_toolkits.mplot3d import Axes3D  # Necesario para los gráficos en 3D

# Crear un gráfico en 3D
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Colorear por especie
species = {'Adelie Penguin (Pygoscelis adeliae)': 'blue',
            'Chinstrap penguin (Pygoscelis antarctica)': 'green',
            'Gentoo penguin (Pygoscelis papua)': 'red'}

for specie, color in species.items():
    subset = train_data[train_data['Species'] == specie]
    ax.scatter(subset['Culmen Length (mm)'], subset['Culmen Depth (mm)'], subset['Flipper Length (mm)'],
                color=color, label=specie, alpha=0.6)

# Etiquetas de los ejes
ax.set_xlabel('Culmen Length (mm)')
ax.set_ylabel('Culmen Depth (mm)')
ax.set_zlabel('Flipper Length (mm)')
ax.set_title('Culmen Length, Culmen Depth y Flipper Length por Especie')

# Agregar leyenda
ax.legend()

# Mostrar gráfico
plt.show()
```

```python
import plotly.express as px

# Cargar el conjunto de entrenamiento
train_data = pd.read_csv('/content/penguins_training.csv')

# Colorear por especie usando Plotly Express
species_map = {'Adelie Penguin (Pygoscelis adeliae)'      : 'blue',
               'Chinstrap penguin (Pygoscelis antarctica)': 'green',
               'Gentoo penguin (Pygoscelis papua)'        : 'red'}

# Crear un gráfico de dispersión 3D interactivo
fig = px.scatter_3d(train_data,
                    x='Culmen Length (mm)',
                    y='Culmen Depth (mm)',
                    z='Flipper Length (mm)',
                    color='Species',  # Plotly asigna colores automáticamente por especie
                    color_discrete_map=species_map,
                    title="Culmen Length, Culmen Depth y Flipper Length por Especie")

# Mostrar el gráfico
fig.show()
```
