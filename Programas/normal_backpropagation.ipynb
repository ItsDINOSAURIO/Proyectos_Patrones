```python
# 1. Función sigmoide (logsig) y su derivada
def sigmoid(n):
  f = 1 / (1 + np.exp(-n))
  return f

def sigmoid_derivada(n):
  f_punto = n * (1 - n)
  return f_punto

def logsig(n):
    f = 1 / (1 + np.exp(-n))
    return f

def logsig_derivada(n):
    f_punto = n * (1 - n)
    return 1 / (1 + np.exp(-n))

# 2. Función lineal (purelin) y su derivada
def purelin(n):
    f = n
    return f

def purelin_derivada(n):
    f_punto = 1
    return f_punto

#Para mejor respuesta en reconocedor de numeros: utilizar softmax, para la última capa,así evitando picos en la derivada, al momento de realizar el cálculo del error
```

```python
# 3. Inicializar los pesos y sesgos de la red neuronal
def initialize_weights(input_size, hidden1_size, hidden2_size, output_size):
    np.random.seed(42)
    weights = {
        'W1': np.random.randn(input_size, hidden1_size)   * np.sqrt(2. / input_size),
        'b1': np.zeros((1, hidden1_size)),
        'W2': np.random.randn(hidden1_size, hidden2_size) * np.sqrt(2. / hidden1_size),
        'b2': np.zeros((1, hidden2_size)),
        'W3': np.random.randn(hidden2_size, output_size)  * np.sqrt(2. / hidden2_size),
        'b3': np.zeros((1, output_size)),
    }
    return weights
```

```python
# 4. Propagación hacia adelante
def forward_propagation(p, weights):
    # Capa oculta 1
    hidden1_input = np.dot(p, weights['W1']) + weights['b1']
    hidden1_output = sigmoid(hidden1_input)

    # Capa oculta 2
    hidden2_input = np.dot(hidden1_output, weights['W2']) + weights['b2']
    hidden2_output = sigmoid(hidden2_input)

    # Capa de salida (purelin - lineal)
    output_input = np.dot(hidden2_output, weights['W3']) + weights['b3']
    output = purelin(output_input)  # Salida lineal

    return hidden1_output, hidden2_output, output
```

```python
# 5. Retropropagación y ajuste de pesos
def train_network(x, y, weights, learning_rate, epochs):
    for epoch in range(epochs):
        total_error = 0

        for p, target in zip(x, y):
            # Paso hacia adelante
            hidden1_output, hidden2_output, output = forward_propagation(p, weights)

            # Calcular error
            error = target - output
            total_error += np.sum(error**2)

            # Retropropagación
            # Gradiente para la salida (capa lineal)
            output_gradient = error * purelin_derivada(output)

            # Gradiente para la capa oculta 2
            hidden2_error = np.dot(output_gradient, weights['W3'].T)
            hidden2_gradient = hidden2_error * sigmoid_derivada(hidden2_output)

            # Gradiente para la capa oculta 1
            hidden1_error = np.dot(hidden2_gradient, weights['W2'].T)
            hidden1_gradient = hidden1_error * sigmoid_derivada(hidden1_output)

            # Actualizar los pesos y sesgos
            weights['W3'] += learning_rate * hidden2_output.reshape(-1, 1) * output_gradient
            weights['b3'] += learning_rate * output_gradient

            weights['W2'] += learning_rate * hidden1_output.reshape(-1, 1) * hidden2_gradient
            weights['b2'] += learning_rate * hidden2_gradient

            weights['W1'] += learning_rate * p.reshape(-1, 1) * hidden1_gradient
            weights['b1'] += learning_rate * hidden1_gradient

        # Monitoreo del error
        if epoch % 100 == 0:
            print(f"Época {epoch}, Error total: {total_error / len(x)}")
```

```python
# 6. Normalizar los datos de entrada
x_train = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])  # Todas las combinaciones binarias/// Vector imagen 
y_train = np.array([[0], [1], [1], [0]])  # Salida deseada de XOR

# Normalizar entradas
x_train = (x_train - np.mean(x_train, axis=0)) / np.std(x_train, axis=0)

# 7. Inicializar pesos
weights = initialize_weights(2, 2, 2, 1) #Para números 128,64,10

# 8. Entrenar la red neuronal
learning_rate = 0.01
epochs = 20000
train_network(x_train, y_train, weights, learning_rate, epochs)
```

```python
# 9. Pruebas con la red entrenada
print("\nResultados de la red neuronal para el problema XOR:")
for x in x_train:
    _, _, output = forward_propagation(x, weights)
    print(f"Entrada: {x}, Salida predicha: {output}")
```
